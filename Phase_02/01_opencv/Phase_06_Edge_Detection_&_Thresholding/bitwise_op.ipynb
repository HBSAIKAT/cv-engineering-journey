```python
"""
1- cv2.bitwise_and(img1, img2)
2- cv2.bitwise_or(img1, img2)
3- cv2.bitwise_not(img)
"""

import cv2
import numpy as np

img1 = np.zeros((300, 300), dtype=np.uint8)
img2 = np.zeros((300, 300), dtype=np.uint8)

cv2.circle(img1, (150, 150), 100, 255, -1)
cv2.rectangle(img2, (100, 100), (250, 250), 255, -1)

and_img = cv2.bitwise_and(img1, img2)
or_img = cv2.bitwise_or(img1, img2)
not_img = cv2.bitwise_not(img1)

cv2.imshow('Image 1', img1)
cv2.imshow('Image 2', img2)
cv2.imshow('AND', and_img)
cv2.imshow('OR', or_img)
cv2.imshow('NOT', not_img)
cv2.waitKey(0)
cv2.destroyAllWindows()

cv2.imwrite('and_image.jpg', and_img)
cv2.imwrite('or_image.jpg', or_img)
cv2.imwrite('not_image.jpg', not_img)
```
