```python
import cv2
import os
import sys
```

```python

# 1. Load classifiers using OpenCV's built-in universal paths
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
eye_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_eye.xml")
smile_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_smile.xml")

```

```python

# 2. Test Image Setup
image_name = "habib.png"

if not os.path.exists(image_name):
    print(f"Error: '{image_name}' not found. Please place an image in this folder named 'test_face.jpg'.")
    sys.exit()
```

```python

# Load and process image
frame = cv2.imread(image_name)
gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
```

```python

# 3. Detect Faces
faces = face_cascade.detectMultiScale(gray, 1.1, 7)

print(f"Detected {len(faces)} face(s). Processing details...")
```

```python

# 4. Process within the face loop (FIXED INDENTATION SCOPE)
for (x, y, w, h) in faces:
    # Draw blue rectangle around face
    cv2.rectangle(frame, (x, y), (x + w, y + h), (255, 0, 0), 2)
    
    # Define Region of Interest (ROI) specifically inside the detected face bounding box
    roi_gray = gray[y : y + h, x : x + w]
    roi_color = frame[y : y + h, x : x + w]

    # 5. Detect Eyes (Using the gray ROI)
    eyes = eye_cascade.detectMultiScale(roi_gray, 1.1, 10)
    if len(eyes) > 0:  # FIXED: Changed from >= 0 to > 0
        cv2.putText(
            frame,
            "Eyes detected",
            (x, y - 35),  # Shifted up slightly
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0, 255, 0),
            2
        )
        # Optional: Draw boxes around the eyes inside the face
        for (ex, ey, ew, eh) in eyes:
            cv2.rectangle(roi_color, (ex, ey), (ex + ew, ey + eh), (0, 255, 0), 1)
    
    # 6. Detect Smile (Using the gray ROI)
    smile = smile_cascade.detectMultiScale(roi_gray, 1.1, 25) # Bumped minNeighbors to 25 to prevent false positives
    if len(smile) > 0:  # FIXED: Changed from >= 0 to > 0
        cv2.putText(
            frame,
            "Smile detected!",
            (x, y - 10),  # FIXED: Changed coordinate so it doesn't overlap "Eyes detected"
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0, 0, 255),
            2
        )
```

```python

# 7. Show Final Result
cv2.imshow("SMART FACE DETECTOR - TEST MODE", frame)
print("Press any key on the image window to exit.")
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.imwrite("face_eye_smile_result.png", frame)  # Optional: Save the result image to disk
```
