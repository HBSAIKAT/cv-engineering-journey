```python
# Line drawing in OpenCV

import cv2

image = cv2.imread('sakib.jpg')
if image is None:
    print('Failed to load image')
else:
    print('Image loaded successfully')

    pt1 = (50,100)
    pt2 = (150, 200)

    color = (255, 0, 0)  
    thickness = 4

    cv2.line(image, pt1, pt2, color, thickness)
    cv2.imshow('Line drawing', image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# Rectangle drawing in OpenCV

import cv2

image = cv2.imread('sakib.jpg')
if image is None:
    print('Failed to load image')
else:
    print('Image loaded successfully')

    pt1 = (50,100)
    pt2 = (150, 200)

    color = (255, 0, 0)  
    thickness = 4

    cv2.rectangle(image, pt1, pt2, color, thickness)
    cv2.imshow('Rectangle drawing', image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# Circle drawing in OpenCV

import cv2

image = cv2.imread('sakib.jpg')
if image is None:
    print('Failed to load image')
else:
    print('Image loaded successfully')

    center = (150,150)
    radius = 50
    color = (255, 0, 0)  
    thickness = 4

    cv2.circle(image, center, radius, color, thickness)
    cv2.imshow('Circle drawing', image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# Adding text to an image in OpenCV

import cv2

image = cv2.imread('sakib.jpg')
if image is None:
    print('Failed to load image')
else:
    print('Image loaded successfully')
    text = 'Hello, OpenCV!'
    org = (10, 40)
    font = cv2.FONT_HERSHEY_SIMPLEX
    fontScale = 1
    color = (255, 0, 0)
    thickness = 4
    
    cv2.putText(image, text, org, font, fontScale, color, thickness)
    cv2.imshow('Text drawing', image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
