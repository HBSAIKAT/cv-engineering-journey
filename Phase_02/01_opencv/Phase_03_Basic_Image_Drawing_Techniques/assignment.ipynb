```python
"""
Task: From user take file input using file handling
After give input give some option, line drawing, circle drawing, rectangle drawing, text writing
if user select line drawing then ask for 2 points and color and thickness and draw line on image
  than output showing
  if he save the image then save the image with new name
  if he not save the image then close the image
if user select circle drawing then ask for center and radius and color and thickness and draw circle on image
  than output showing
  if he save the image then save the image with new name
  if he not save the image then close the image
if user select rectangle drawing then ask for 2 points and color and thickness and draw rectangle on image
  than output showing
  if he save the image then save the image with new name
  if he not save the image then close the image
if user select text writing then ask for text and position and color and thickness and write text on image
  than output showing
  if he save the image then save the image with new name
  if he not save the image then close the image
"""
```

```python
import cv2
file_path = input("Enter the image file path: ")
image = cv2.imread(file_path)
if image is None:
    print('Error loading image. Please check the file path and try again.')
else:
    print('Image loaded successfully')
    input_option = input("Select an option: 1 for Line Drawing, 2 for Circle Drawing, 3 for Rectangle Drawing, 4 for Text Writing: ")
    if input_option == '1':
        pt1 = tuple(map(int, input("Enter the first point (x1, y1) separated by space: ").split()))
        pt2 = tuple(map(int, input("Enter the second point (x2, y2) separated by space: ").split()))
        color = tuple(map(int, input("Enter the color (B, G, R) separated by space: ").split()))
        thickness = int(input("Enter the thickness: "))
        cv2.line(image, pt1, pt2, color, thickness)
        cv2.imshow('Line drawing', image)
        cv2.waitKey(0)
        save_option = input("Do you want to save the image? (y/n): ")
        if save_option.lower() == 'y':
            new_file_name = input("Enter the new file name (with extension): ")
            cv2.imwrite(new_file_name, image)
            print(f"Image saved as {new_file_name}")

    elif input_option == '2':
        center = tuple(map(int, input("Enter the center point (x, y) separated by space: ").split()))
        radius = int(input("Enter the radius: "))
        color = tuple(map(int, input("Enter the color (B, G, R) separated by space: ").split()))
        thickness = int(input("Enter the thickness: "))
        cv2.circle(image, center, radius, color, thickness)
        cv2.imshow('Circle drawing', image)
        cv2.waitKey(0)
        save_option = input("Do you want to save the image? (y/n): ")
        if save_option.lower() == 'y':
            new_file_name = input("Enter the new file name (with extension): ")
            cv2.imwrite(new_file_name, image)
            print(f"Image saved as {new_file_name}")

    elif input_option == '3':
        pt1 = tuple(map(int, input("Enter the first point (x1, y1) separated by space: ").split()))
        pt2 = tuple(map(int, input("Enter the second point (x2, y2) separated by space: ").split()))
        color = tuple(map(int, input("Enter the color (B, G, R) separated by space: ").split()))
        thickness = int(input("Enter the thickness: "))
        cv2.rectangle(image, pt1, pt2, color, thickness)
        cv2.imshow('Rectangle drawing', image)
        cv2.waitKey(0)
        save_option = input("Do you want to save the image? (y/n): ")
        if save_option.lower() == 'y':
            new_file_name = input("Enter the new file name (with extension): ")
            cv2.imwrite(new_file_name, image)
            print(f"Image saved as {new_file_name}")

    elif input_option == '4':
        pt = tuple(map(int, input("Enter the point (x, y) separated by space: ").split()))
        text = input("Enter the text: ")
        color = tuple(map(int, input("Enter the color (B, G, R) separated by space: ").split()))
        thickness = int(input("Enter the thickness: "))
        font_scale = float(input("Enter the font scale: "))
        cv2.putText(image, text, pt, cv2.FONT_HERSHEY_SIMPLEX, font_scale, color, thickness)
        cv2.imshow('Text writing', image)
        cv2.waitKey(0)
        save_option = input("Do you want to save the image? (y/n): ")
        if save_option.lower() == 'y':
            new_file_name = input("Enter the new file name (with extension): ")
            cv2.imwrite(new_file_name, image)
            print(f"Image saved as {new_file_name}")

    else:
        print("Invalid option selected.")
```
