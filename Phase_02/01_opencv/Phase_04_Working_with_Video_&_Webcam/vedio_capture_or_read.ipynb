```python
import cv2
cap = cv2.VideoCapture(0) 
# start video capture from webcam (0 is the default webcam)

while True:
    ret, frame = cap.read() # ret = true/false, frame = image
    if not ret:
        print('Failed to capture video')
        break
    cv2.imshow('Webcam feed', frame)

    if cv2.waitKey(1) & 0xFF == ord('q'): # 113 == 113 True if 'q' is pressed
        print("Quit signal received. Exiting...")
        break
cap.release()
cv2.destroyAllWindows()

"""
waitKey(1) means wait for 1 millisecond for a key press. 
If 'q' is pressed, it will break the loop and exit the program.
release() is called to free the webcam resource, 
and destroyAllWindows() is called to close any OpenCV windows that were opened.
cap.read() reads a frame from the webcam. 
It returns a boolean (ret) indicating if the frame was read successfully, 
and the frame itself (frame).
0xFF == ord('q') is a common way to check if the 'q' key was pressed.
"""
```
